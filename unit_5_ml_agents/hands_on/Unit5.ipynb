### Define the SnowballTarget config file
- In ML-Agents, you define the **training hyperparameters into config.yaml files.**

There are multiple hyperparameters. To know them better, you should check for each explanation with [the documentation](https://github.com/Unity-Technologies/ml-agents/blob/release_20_docs/docs/Training-Configuration-File.md)


So you need to create a `SnowballTarget.yaml` config file in ./content/ml-agents/config/ppo/

We'll give you here a first version of this config (to copy and paste into your `SnowballTarget.yaml file`), **but you should modify it**.

```
behaviors:
  SnowballTarget:
    trainer_type: ppo
    summary_freq: 10000
    keep_checkpoints: 10
    checkpoint_interval: 50000
    max_steps: 200000
    time_horizon: 64
    threaded: false
    hyperparameters:
      learning_rate: 0.0003
      learning_rate_schedule: linear
      batch_size: 128
      buffer_size: 2048
      beta: 0.005
      epsilon: 0.2
      lambd: 0.95
      num_epoch: 3
    network_settings:
      normalize: false
      hidden_units: 256
      num_layers: 2
      vis_encode_type: simple
    reward_signals:
      extrinsic:
        gamma: 0.99
        strength: 1.0
```

### Create the SnowballTarget config file

We need to create the `SnowballTarget.yaml` config file in `./content/ml-agents/config/ppo/` with the provided hyperparameters.

````python
import os

# Define the directory path
config_dir = './config/ppo/'

# Create the directory if it doesn't exist
os.makedirs(config_dir, exist_ok=True)

# Define the YAML content
yaml_content = '''
behaviors:
  SnowballTarget:
    trainer_type: ppo
    summary_freq: 10000
    keep_checkpoints: 10
    checkpoint_interval: 50000
    max_steps: 200000
    time_horizon: 64
    threaded: false
    hyperparameters:
      learning_rate: 0.0003
      learning_rate_schedule: linear
      batch_size: 128
      buffer_size: 2048
      beta: 0.005
      epsilon: 0.2
      lambd: 0.95
      num_epoch: 3
    network_settings:
      normalize: false
      hidden_units: 256
      num_layers: 2
      vis_encode_type: simple
    reward_signals:
      extrinsic:
        gamma: 0.99
        strength: 1.0
'''

# Define the file path
file_path = os.path.join(config_dir, 'SnowballTarget.yaml')

# Write the content to the file
with open(file_path, 'w') as f:
    f.write(yaml_content)

print(f"Created {file_path}")
````

### Fix: `FileNotFoundError: results/PyramidsTraining/configuration.yaml`

The `mlagents-push-to-hf` utility expects a `configuration.yaml` file to be present in the `local-dir` (i.e., `./results/PyramidsTraining`) for generating the model card. Since the `PyramidsRND.yaml` was used directly from the `ml-agents/config/ppo/` directory, it needs to be explicitly copied and renamed to `configuration.yaml` in the results folder.

````python
import os
import shutil

# Define source and destination paths
source_config_path = './config/ppo/PyramidsRND.yaml'
dest_dir = './results/PyramidsTraining/'
dest_config_path = os.path.join(dest_dir, 'configuration.yaml')

# Ensure the destination directory exists
os.makedirs(dest_dir, exist_ok=True)

# Copy and rename the config file
shutil.copyfile(source_config_path, dest_config_path)

print(f"Copied '{source_config_path}' to '{dest_config_path}'")
````

````python
import os
import shutil
import glob

src = "./results/Pyramids Training"
dst = "./results/PyramidsTraining"

assert os.path.exists(src), "No encuentro la carpeta original de entrenamiento: ./results/Pyramids Training"

if os.path.exists(dst):
    shutil.rmtree(dst)

shutil.copytree(src, dst)

shutil.copyfile(
    "./config/ppo/PyramidsRND.yaml",
    os.path.join(dst, "configuration.yaml")
)

print("Archivos ONNX encontrados:")
print(glob.glob(dst + "/**/*.onnx", recursive=True))

print("Archivos PT encontrados:")
print(glob.glob(dst + "/**/*.pt", recursive=True))
````
